## Recent 6-months

```python
df_reviews = pd.read_feather('data/appliances_meta_012023_062023 v1.0.0.ftr')
```

```python
col = "images"   # <- change to your real column name

def to_list(x):
    if isinstance(x, (list, tuple)): 
        return list(x)
    if isinstance(x, np.ndarray): 
        return x.tolist()
    if pd.isna(x): 
        return []
    return [x]  # single dict or scalar

def extract_mediums(seq):
    out = []
    for item in seq:
        if isinstance(item, dict) and item.get("attachment_type") == "IMAGE":
            url = item.get("medium_image_url")
            if url:
                out.append(url)
    return out

seqs = df_reviews[col].apply(to_list)
medium_lists = seqs.apply(extract_mediums)

# 1) If you want a single column with the first medium image (or NaN):
df_reviews["medium_image_url"] = medium_lists.str[0]

# 2) If you want one column per image:
exp = pd.DataFrame(medium_lists.tolist(), index=df_reviews.index)
exp.columns = [f"medium_image_url_{i+1}" for i in range(exp.shape[1])]
df_reviews_imgs = df_reviews.drop(columns=[col]).join(exp)
```

```python
''' Slice products with more than 2 reviews -> to calculate inter-review time '''
df_rv_vc = df_reviews_imgs['parent_asin'].value_counts()
df_reviews_mt2rvs = df_reviews_imgs[df_reviews_imgs['parent_asin'].isin(df_rv_vc[df_rv_vc>=2].index)]
```

```python
''' 11,701 products with average 9 reviews'''
df_reviews_mt2rvs['parent_asin'].nunique(), df_reviews_mt2rvs['parent_asin'].value_counts().mean()
```

```python
df_reviews_mt2rvs.head()
```

```python
''' 91.79% of reviews do not have review images '''
len(df_reviews_mt2rvs[df_reviews_mt2rvs['medium_image_url'].isnull()]) / len(df_reviews_mt2rvs)
```

# Merge Metadata

```python
df_meta = pd.read_feather('data/1_raw/appliances_meta v1.0.0.ftr')[['parent_asin', 'main_category', 'average_rating', 'rating_number', 'price', 'store', 'categories',]]
df_reviews = pd.read_feather('data/appliances_reviews_012023_062023 v1.3.0.ftr')
```

```python
df_meta['categories'] = df_meta['categories'].str[1]
```

```python
df_meta.head()
```

```python
df_reviews_mrgMeta = df_reviews.merge(df_meta, on='parent_asin', how='left')
```

```python
col_txt = [col for col in df_reviews_mrgMeta.columns if 'MPN' in col]
col_img = [col for col in df_reviews_mrgMeta.columns if 'ResNet' in col]
col_etc = set(list(df_reviews_mrgMeta.columns)) - set(col_txt) - set(col_img)
```

```python
df_reviews_mrgMeta_org = df_reviews_mrgMeta[['review_id', 'rating', 'title', 'text', 'asin', 'parent_asin', 'user_id', 'timestamp', 'helpful_vote', 'verified_purchase', 'text_len', 'text_words', 
                                             'inter_review_time', 'main_category', 'average_rating', 'rating_number', 'price', 'store', 'categories'] + col_txt + col_img]
```
